## Sistema de Pronóstico Financiero de Grupo Éxito

### Objetivo

Construir un dataset histórico trimestral a partir de información financiera oficial de Grupo Éxito para analizar el comportamiento financiero y desarrollar modelos de pronóstico.

Grupo Éxito – Información Financiera

Periodo inicial de análisis: 2015–2025.

### Variables objetivo

- Ingresos operacionales

- EBITDA
- Resultado neto

```python
# Librerias

import pandas as pd
import os
from pathlib import Path

```

```python
# Ruta del proyecto

PROJECT_ROOT = Path("..")

BRONZE_PATH = PROJECT_ROOT / "data" / "Bronze"
SILVER_PATH = PROJECT_ROOT / "data" / "Silver"
GOLD_PATH = PROJECT_ROOT / "data" / "Gold"

print("Bronze", BRONZE_PATH)
print("Silver", SILVER_PATH)
print("Gold", GOLD_PATH)
```

```python
df_financiero = pd.read_excel(BRONZE_PATH / "informe_resultados_exito.xlsx")

df_financiero.head()
```
